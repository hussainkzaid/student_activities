# Key Findings & Insights (EDA Summary)

**Dataset:** `StudentPerformanceFactors-selected-columns.csv` — 6,607 rows x 10 columns, no duplicates, no missing values.

### Data prep performed
- **Ordinal encoding** (Low=0, Medium=1, High=2): `Parental_Involvement`, `Access_to_Resources`, `Motivation_Level`
- **Label encoding** (binary): `Extracurricular_Activities`, `Internet_Access`
- **Min-Max scaling** (0-1): `Hours_Studied`, `Attendance`, `Sleep_Hours`, `Previous_Scores`, `Tutoring_Sessions`

### Relationship findings
- **Parental involvement, access to resources, and motivation level** all show a clear **positive step-up** in average `Previous_Scores` from Low → Medium → High. These are the strongest and most consistent categorical signals in the data.
- **Attendance vs. Previous_Scores**: essentially **no linear correlation** (r ≈ -0.02). The scatter plot shows a wide, flat spread with no visible trend.
- **Hours_Studied vs. Previous_Scores**: essentially **no linear correlation** (r ≈ 0.02). Scores stay roughly flat (~74-78) across almost the whole study-hours range.
- **Sleep_Hours vs. Previous_Scores**: average score is nearly flat (~74-76) across 4-10 hours of sleep — no meaningful relationship.
- **Internet_Access vs. Previous_Scores**: average scores for Yes (75.1) and No (74.9) are nearly identical — negligible effect despite the strong class imbalance.
- **Correlation heatmap** (scaled numeric features `Hours_Studied`, `Attendance`, `Sleep_Hours`, `Previous_Scores`, `Tutoring_Sessions`): all off-diagonal correlations are close to 0, confirming these continuous features are largely independent of each other and of `Previous_Scores`.
