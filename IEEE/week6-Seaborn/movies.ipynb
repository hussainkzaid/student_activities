```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style='darkgrid')
```

```python
movies = pd.read_csv(r"C:\Users\hussa\OneDrive\Desktop\IEEE_task\movies\mymoviedb.csv", engine="python")
movies
```

```python
movies.info()
```

```python
movies['Release_Date'] = pd.to_datetime(movies['Release_Date'], errors='coerce')
movies['Release_Date']
```

```python
movies['Vote_Average'] = pd.to_numeric(movies['Vote_Average'],errors='coerce')
movies['Vote_Count'] = pd.to_numeric(movies['Vote_Count'],errors='coerce')
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies[movies.isnull()]
```

```python
movies.dropna(inplace=True)
```

```python
pop_movies = movies.sort_values(by='Popularity', ascending=False).head(10)
pop_movies
```

```python
#What is the movie with the highest Popularity rate? And what kind is it?
colors = list(plt.cm.Greens(np.linspace(1, 0.3, len(pop_movies))))
sns.barplot(x=pop_movies['Popularity'], y=pop_movies['Title'], hue=pop_movies['Title'], palette=colors)
plt.title('Movies with the highest Popularity rate')
```

```python
#what kind is it?
colors = list(plt.cm.Greens(np.linspace(1, 0, len(pop_movies))))
sns.barplot(x=pop_movies['Popularity'], y=pop_movies['Genre'], hue=pop_movies['Genre'], palette='hot')
plt.title("Genres of the highest Popularity rate movies")
```

```python
movies['Year'] = movies['Release_Date'].dt.year
movies['Year']
```

```python
#What year was the largest number of films produced?
year_prod = movies.groupby('Year')['Title'].count()
sns.lineplot(x=year_prod.index, y=year_prod)
plt.ylabel('Number of Films')
plt.title('Number of Films Produced by Year')
```

```python
genres = movies['Genre'].str.split(', ')
genres = genres.explode()
genres
```

```python
genres_count = genres.value_counts()
genres_count
```

```python
plt.colormaps()
```

```python
#•What is most of the film genre?
sns.barplot(x=genres_count, y=genres_count.index, hue=genres_count.index, palette='summer')
plt.title('number of films by genre')
```

```python
#What kind of movies take the highest rating?
genre_rating = movies[['Genre', 'Vote_Average']].copy()
genre_rating['Genre'] = movies['Genre'].str.split(', ')
genre_rating = genre_rating.explode('Genre')
genre_rating
```

```python
genre_rating = genre_rating.groupby('Genre')['Vote_Average'].mean().sort_values(ascending = False)
genre_rating
```

```python
sns.barplot(x=genre_rating, y=genre_rating.index, hue=genre_rating, legend=False)
plt.title('Vote Average by genre')
```

```python
#Which movie is more popular? Which movie is less popular?
genre_popular = movies[['Genre', 'Popularity']].copy()
genre_popular['Genre'] = movies['Genre'].str.split(', ')
genre_popular = genre_popular.explode('Genre')
genre_popular
```

```python
genre_popular = genre_popular.groupby('Genre')['Popularity'].mean().sort_values(ascending=False)
genre_popular
```

```python
sns.barplot(x=genre_popular, y=genre_popular.index, hue=genre_popular, legend=False, palette='magma_r')
```

```python
#How has the average movie rating changed over the years?
sns.lineplot(x=movies['Year'], y=movies['Vote_Average'], color='red')
plt.title('Average movie rating over years')
```

```python
lang_pop = movies.groupby('Original_Language')['Popularity'].sum().sort_values(ascending=False).head(10)
lang_pop
```

```python
#movies language with the highest popularity
sns.barplot(x=lang_pop.index, y=lang_pop, color='orange')
plt.title('popularity by movies language')
```

```python
lang_vote = movies.groupby('Original_Language')['Vote_Average'].mean().sort_values(ascending=False).head(5)
lang_vote
```

```python
#Is there a relationship between popularity and rating?
sns.scatterplot(x=movies['Vote_Average'], y=movies['Popularity'], alpha=0.5)
plt.xlabel('Average Rating')
plt.ylabel('Popularity')
plt.title('Relationship Between Movie Rating and Popularity')
```
