4- Add the following columns:

a. `Passed`

Rules:
- Grade >= 60 → `"Yes"`
- Otherwise → `"No"`

b. `Grade_Level`

Rules:
- Grade >= 90 → `"Excellent"`
- Grade >= 80 → `"Good"`
- Otherwise → `"Average"`

Then change the department of `"Omar"` from `"CS"` to `"AI"` without recreating the DataFrame.

````python
def passed(grade):
  if grade >= 60:
    return 'Yes'
  else:
    return 'No'

def grade_level(grade):
  if grade >= 90:
    return 'Excellent'
  elif grade >= 80:
    return 'Good'
  else:
    return 'Average'

df['Passed'] = df['Grade'].apply(passed)

df['Grade_Level'] = df['Grade'].apply(grade_level)

df.loc[df['Name'] == 'Omar' , 'Department'] = 'AI'

df
````

7- Create the following DataFrame and store it in `df_missing`


```
data = {
    "Name": ["Ali", "Sara", None, "Lina"],
    "Age": [20, None, 19, 22],
    "Grade": [85, 92, 78, None]
}
```

Then:

- Display missing values in each column  
- Display total missing values  
- Fill missing names with `"Unknown"`  
- Fill numeric missing values with the column average  

````python
data = {
    "Name": ["Ali", "Sara", None, "Lina"],
    "Age": [20, None, 19, 22],
    "Grade": [85, 92, 78, None]
}

df_missing = pd.DataFrame(data)

print(df_missing.isnull())

print('##############################################')

print(df_missing.isnull().sum())

print('##############################################')

df_missing['Name'] = df_missing['Name'].fillna('Unknown')

df_missing['Age'] = df_missing['Age'].fillna(df_missing['Age'].mean())

df_missing['Grade'] = df_missing['Grade'].fillna(df_missing['Grade'].mean())

df_missing
````

8- Create the following DataFrames:


```
students_info = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Omar", "Lina"],
    "Department": ["CS", "AI", "CS", "DS"]
})

students_grades = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Grade": [85, 92, 78, 95]
})
```
Combine both DataFrames into one DataFrame using the `ID` column.


````python
students_info = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Omar", "Lina"],
    "Department": ["CS", "AI", "CS", "DS"]
})

students_grades = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Grade": [85, 92, 78, 95]
})

students_DataFrame = pd.merge(students_info , students_grades , how = 'outer' , on = 'ID')

students_DataFrame
````

new_students = pd.DataFrame({
