```python
#Load the dataset using Pandas.
import pandas as pd

books = pd.read_csv("books.csv")
borrowings = pd.read_csv("borrowings.csv")
members = pd.read_csv("members.csv")
```

```python
#Explore the dataset by displaying enough information to understand its structure before beginning the analysis.
print("**************** FOR BOOKS ****************")
print(books)
print(books.info())

print("**************** FOR BORROWINGS ****************")
print(borrowings)
print(borrowings.info())

print("**************** FOR MEMBERS ****************")
print(members)
print(members.info())
```

```python
# the probability of randomly selecting A borrowed book from a specific genre
total_books = len(books)
book_genres = books['Genre'].unique()
print(f"the generies is {book_genres}")
for genre in book_genres :
  total_specific = len(books[books["Genre"] == genre])
  probability = (total_specific / total_books) * 100
  print(f"the probability of randomly selecting A borrowed book from a {genre} genre is {probability}%")

print("\n\n")

# the probability of randomly selecting A member belonging to a specific membership type.
total_members = len(members)
members_type = members['MembershipType'].unique()
print(f"the memberships is {members_type}")
for member_type in members_type:
  total_specific = len(members[members['MembershipType'] == member_type])
  probability =  (total_specific / total_members) * 100
  print(f"the probability of randomly selecting A member belonging to a {member_type} membership type is {probability:0.1f}%")

print("\n\n")

#the probability of randomly selecting A borrowing record that satisfies two conditions of your choice.
#I set two conditions -> borrowed in after 2024 and returned before 2026
total_borrowing = len(borrowings)
borrowings['BorrowDate'] = pd.to_datetime(borrowings['BorrowDate'])
borrowings['ReturnDate'] = pd.to_datetime(borrowings['ReturnDate'])
total_specific = len(borrowings[(borrowings['BorrowDate'].dt.year > 2024) & (borrowings['ReturnDate'].dt.year  < 2026)])
probability = (total_specific / total_borrowing) * 100
print(f"the probability of randomly selecting A borrowing record that borrowed in after 2024 and returned before 2026 is {probability}")
```

```python
borrowings_books_table = pd.merge(books, borrowings, how = 'inner', on = 'BookID')
borrowings_members_table = pd.merge(borrowings, members, how='inner', on='MemberID')
borrowings_members_books_table = pd.merge(books, borrowings_members_table, how='inner', on='BookID')
```

```python
#Using Pandas, answer the following:
#Find the category with the highest and lowest number of borrowings.
category_borrowings = borrowings_books_table.groupby('Genre')['BorrowID'].count().sort_values(ascending=False)
print(category_borrowings)
print(f"So the category with the highest number of borrowings is {category_borrowings.idxmax()} with {category_borrowings.max()} borrowed books and the lowest is {category_borrowings.idxmin()} with {category_borrowings.min()} borrowed books.")
print("***********************************")

#Determine which members borrow books most frequently.
print(borrowings_members_table.groupby('Name')['BorrowID'].count().sort_values(ascending=False))

print("***********************************")

#Compare borrowing activity between different membership types.
print(borrowings_members_table.groupby('MembershipType')['BorrowID'].count().sort_values(ascending=False))

print("***********************************")
```

```python
#Analyze how borrowing changes across publication years.
print(borrowings_books_table.groupby('PublicationYear')['BorrowID'].count())

print("***********************************")

#Identify the most popular authors.
author_borrowing = borrowings_books_table.groupby('Author')['BorrowID'].count().sort_values(ascending=False)
print(author_borrowing)
print(f'So the most popular authors is {author_borrowing.idxmax()} with {author_borrowing.max()} borrowed books')
```

```python
#Find the books that have never been borrowed.
merge = pd.merge(books, borrowings, how = 'left', on = 'BookID')
book_nborrowwed = merge[merge['BorrowID'].isna()]
print(book_nborrowwed['Title'])

print("***********************************")

#Compare the average borrowing duration across genres.
borrowings_books_table['Borrowing Duration'] = borrowings_books_table['ReturnDate'] - borrowings_books_table['BorrowDate']
print(borrowings_books_table.groupby('Genre')['Borrowing Duration'].mean().sort_values(ascending=False))
```

```python
#Show the distribution of books across genres.
print(books.groupby('Genre')['BookID'].count().sort_values(ascending = False))

print("***********************************")

#Analyze monthly borrowing trends (if borrowing dates are available).
borrowings_books_table['Borrowing Month'] = borrowings_books_table['BorrowDate'].dt.month
print(borrowings_books_table.groupby('Borrowing Month')['BorrowID'].count())

print("***********************************")

#Present three additional insights of your choice
#Which age group borrows the most books
print(borrowings_members_table.groupby('Age')['BorrowID'].count().sort_values(ascending = False).head())

print("***********************************")

#Average number of pages by genre
print(books.groupby('Genre')['Pages'].mean().astype(int).sort_values(ascending = False))

print("***********************************")

#Which genre has the most total copies available
print(books.groupby('Genre')['CopiesAvailable'].sum().sort_values(ascending = False))
```

## Final Summary: Key Insights from the Library Management Analysis

**Dataset overview:** 400 books, 2,000 borrowing records, and 150 members across 4 membership types and 10 genres.

### Borrowing patterns
- **Mystery is the most-borrowed genre by far** (512 borrowings), followed by Data Science (328) and Fantasy (316). **Romance is the least borrowed** (81), despite not being the smallest category in the catalog — an early sign that catalog size and borrowing demand don't move together.
- **Students drive the vast majority of activity**: 1,197 of 2,000 borrowings (~60%), consistent with Students also being 60% of the membership base. Public members (20% of members) account for 415 borrowings, Faculty (14.7%) account for 289, and Premium members (5.3%, the smallest group) account for just 99 — roughly proportional to their share of membership, suggesting no single membership type borrows disproportionately more *per member* than expected from its size alone.
- **Borrowing activity is highly concentrated among a few power users**: Member 78 alone borrowed 31 books, compared to a long tail of members who borrowed only 4–6 books. This kind of concentration is worth flagging for engagement programs.
- **Alice Brown is the most borrowed author** (191 borrowings), narrowly ahead of David Clark (178) and Sarah Mitchell (170); the spread across the top 5 authors is fairly tight, indicating broad author appeal rather than reliance on one star author.
- **Borrowing has a clear seasonal dip**: monthly borrowings peak in March (218) and decline steadily through the year to a low in September (115) — roughly a 47% drop from peak to trough. This points to a strong case for targeted promotions or events in the slower autumn months.

### Catalog and inventory insights
- **23 of 400 books (~5.75%) have never been borrowed**, including titles like *Book 12*, *Book 31*, and *Book 213*. These are candidates for promotion, repositioning, or eventual de-acquisition.
- **Genre popularity and catalog size are mismatched.** Business has the largest catalog (52 books) but ranks only 5th in borrowings (121), while Mystery has a mid-sized catalog (43 books) yet generates the most borrowings (512) — over 4x more demand per book than Business. This is the clearest actionable insight in the dataset: **the library may be over-stocked in Business/Self Help and under-stocked in Mystery/Data Science relative to demand.**
- **Borrowing duration is fairly uniform across genres** (15.8–18.1 days), with Biography borrowed longest and Fantasy returned fastest — a relatively narrow ~2-day spread suggests loan length is driven more by library policy/due dates than by genre-specific reading habits.
- Average book length ranges from 394 pages (Romance) to 461 pages (Mystery) — genre doesn't strongly predict book length, and page count alone doesn't explain why Mystery dominates borrowings.

### Member behavior
- **Borrowing is fairly evenly spread across age groups**, with the busiest single ages (53, 38, 24, 59, 31) each contributing similar counts — there's no strong evidence that one age bracket dominates usage.
- The probability calculations show genre selection probabilities ranging narrowly from 8% to 13%, meaning **no single genre dominates the catalog composition** — the borrowing skew toward Mystery is a demand effect, not a supply effect.

### Recommendations for the library
1. **Rebalance acquisitions**: shift future purchasing toward Mystery and Data Science, and slow down acquisitions in Business and Self Help, where supply already outpaces demand.
2. **Run a "rediscover" campaign** for the 23 never-borrowed titles (featured displays, staff picks, recommendation engine boosts) before considering removing them from the catalog.
3. **Target retention efforts at Student members**, since they generate the majority of borrowings — small improvements in their experience would have an outsized impact on overall library usage.
4. **Plan for the September–October slowdown** with seasonal promotions, reading challenges, or events to flatten the year-over-year dip in borrowing activity.
5. **Investigate and recognize top borrowers** (e.g., Member 78) — understanding what drives their high engagement could inform strategies to lift the long tail of lower-activity members.
